```python
%%writefile main.cpp
#include <iostream>
#include <iomanip>
#include <cassert>
#include "Matrix.h"

using namespace std;

int main() {
    int m, n;
    cout << "Enter dimensions of matrix: ";
    cin >> m >> n;
    Matrix a(m, n);
    cout << "Enter elements of matrix:" << endl;
    cin >> a;
    Matrix a0 = a;
    Matrix aOrig = a;


    cout << fixed << setprecision(6);
    cout << "Initial matrix:" << endl;
    cout << a;

    int rank = a.gauss();
    cout << "Rank of matrix = " << rank << endl;
    cout << "Echelon form:" << endl;
    cout << a;

    if (a.rows() == a.cols()) {
        double d = a0.determinant();
        cout << "Determinant = " << d << endl;
    }

    int rank1 = a0.rref();
    assert(rank == rank1);
    cout << "Reduced row echelon form:" << endl;
    cout << a0;

    // Демонстрация обратной матрицы и решения СЛАУ
    if (aOrig.rows() == aOrig.cols() && rank1 == aOrig.rows()) {
        Matrix inv = aOrig.inverse();
        cout << "Inverse matrix:" << endl;
        cout << inv;

        // проверка: A * A^{-1} == E
        Matrix E = aOrig * inv;
        cout << "A * A^{-1} =" << endl;
        cout << E;

        // решение СЛАУ Ax = b
        vector<double> b(a0.rows()), x(a0.rows());
        cout << "Enter right-hand side vector (" << a0.rows() << " values): ";
        for (int i = 0; i < a0.rows(); ++i) cin >> b[i];
        aOrig.solveLinearSystem(b.data(), x.data());
        cout << "Solution:" << endl;
        for (int i = 0; i < a0.rows(); ++i)
            cout << x[i] << " ";
        cout << endl;
    }
        // Читаем две новые матрицы и применяем операции
    int p, q;
    cout << "Enter dimensions for A2 and B2 (p q): ";
    cin >> p >> q;

    Matrix A2(p, q), B2(p, q);
    cout << "Enter elements of A2:\n";
    cin >> A2;
    cout << "Enter elements of B2:\n";
    cin >> B2;

    cout << "A2 =\n" << A2;
    cout << "B2 =\n" << B2;

    cout << "A2 + B2 =\n" << (A2 + B2);
    cout << "A2 - B2 =\n" << (A2 - B2);
    cout << "A2 * B2 =\n" << (A2 * B2);

    return 0;
}
```
