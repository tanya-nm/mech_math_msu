```python
#task3
%%writefile Matrix.h
#ifndef MATRIX_H
#define MATRIX_H

#include <stdexcept>
#include <cmath>
#include <vector>
#include <iostream>

class Matrix {
private:
    int m;  // Number of rows
    int n;  // Number of columns
    std::vector<double> elems;
public:
    Matrix(int num_rows = 1, int num_cols = 1):
        m(num_rows),
        n(num_cols),
        elems(m*n)
    {}

    Matrix(const Matrix& a):
        m(a.m), n(a.n), elems(a.elems)
    {}

    Matrix(Matrix&& a) noexcept:
        m(a.m), n(a.n), elems(std::move(a.elems))
    {
        a.m = 0; a.n = 0;
    }

    Matrix& operator=(const Matrix& a) {
        if (this != &a) {
            m = a.m; n = a.n; elems = a.elems;
        }
        return *this;
    }

    Matrix& operator=(Matrix&& a) noexcept {
        if (this != &a) {
            m = a.m; n = a.n;
            elems = std::move(a.elems);
            a.m = 0; a.n = 0;
        }
        return *this;
    }

    int rows() const { return m; }
    int columns() const { return n; }
    int cols() const { return n; }

    const double* operator[](int i) const {
        return elems.data() + i*n;
    }
    double* operator[](int i) {
        return elems.data() + i*n;
    }

    double at(int i, int j) const {
        if (i < 0 || i >= m || j < 0 || j >= n)
            throw std::out_of_range("Matrix index out of bounds");
        return elems[i*n + j];
    }
    double& at(int i, int j) {
        if (i < 0 || i >= m || j < 0 || j >= n)
            throw std::out_of_range("Matrix index out of bounds");
        return elems[i*n + j];
    }

    // Арифметические операции
    Matrix operator+(const Matrix& b) const;
    Matrix operator-(const Matrix& b) const;
    Matrix operator*(const Matrix& b) const;

    // Приведение к ступенчатому виду (метод Гаусса).
    // Возвращает ранг матрицы.
    int gauss(double eps = 1e-12);

    // Приведение к улучшенному ступенчатому виду (RREF).
    // Возвращает ранг матрицы.
    int rref(double eps = 1e-12);

    // Определитель (только для квадратной матрицы).
    double determinant(double eps = 1e-12) const;

    // Обратная матрица (методом элементарных преобразований).
    Matrix inverse(double eps = 1e-12) const;

    // Решение СЛАУ Ax = b для невырожденной A.
    // freeTerms — массив длины m, solution — массив длины n.
    void solveLinearSystem(const double* freeTerms,
                           double* solution,
                           double eps = 1e-12) const;

    // Дружественные функции ввода/вывода
    friend std::ostream& operator<<(std::ostream& s, const Matrix& a);
    friend std::istream& operator>>(std::istream& s, Matrix& a);
};

std::ostream& operator<<(std::ostream& s, const Matrix& a);
std::istream& operator>>(std::istream& s, Matrix& a);

#endif
```
